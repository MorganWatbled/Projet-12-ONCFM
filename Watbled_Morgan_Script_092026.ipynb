```python
import pandas as pd
import joblib

# Chargement du modèle et du scaler
modele = joblib.load("regression_logistique_billets.pkl")
scaler = joblib.load("scaler.pkl")

# Chargement des billets
billets = pd.read_csv("data/billets_test.csv")

variables = [
    "diagonal",
    "height_left",
    "height_right",
    "margin_low",
    "margin_up",
    "length"
]

X_billets = billets[variables]

# Standardisation
X_billets_scaled = scaler.transform(X_billets)

# Prédictions
predictions = modele.predict(X_billets_scaled)

# Probabilités
probabilites = modele.predict_proba(X_billets_scaled)

billets["prediction"] = predictions

billets["proba_faux"] = probabilites[:, 0] * 100
billets["proba_authentique"] = probabilites[:, 1] * 100

billets["prediction"] = billets["prediction"].map({
    0: "Billet faux",
    1: "Billet authentique"
})

print(
    billets[
        ["id", "prediction", "proba_faux", "proba_authentique"]
    ]
)
billets.to_csv("resultats_predictions.csv", index=False)
```
